```python
#Conocido el radio de la base y la altura de un cilindro, calcular su volumen
radio=float(input("ingese el radio."))
altura=float(input("ingrese la altura."))
volumen=3.1416*((radio**2)*altura)
print(f"el volumen es {volumen}")
```

**Especificacion**
*Interfaz*:
entrada: radio, altura
salida: volumen
*Efecto:*
condiciones de entrada: radio y altura son numeros reales
Efecto producido: se calcula el volumen
*Algoritmo:*
volumen de un cilidro
*Variables: *radio, altura
*principio*:
leer(radio)
leer(altura)
calcular volumen= 3.1416 * radio**2 * altura
escribir(volumen)
